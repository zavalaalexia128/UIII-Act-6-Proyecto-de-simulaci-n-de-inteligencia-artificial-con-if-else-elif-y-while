### Clase SIEFORE (Sociedad de inversión en fondos para el retiro)

```python
print("==========================================")
print(" SISTEMA INTELIGENTE SIEFORE ")
print(" Sociedad de inversion para el retiro ")
print("==========================================")

continuar = "si"

while continuar == "si":

    print("\nResponde las siguientes preguntas\n")

    edad = int(input("Ingresa tu edad: "))
    ahorro = float(input("Ingresa tu ahorro actual: "))
    riesgo = input("Nivel de riesgo (bajo/medio/alto): ").lower()
    retiro = int(input("¿En cuantos años planeas retirarte?: "))

    print("\nAnalizando informacion...")
    print("----------------------------------")
    if edad < 30:
        if riesgo == "alto":
            print("IA SIEFORE:")
            print("Se recomienda inversion agresiva.")
            print("Mayor rendimiento a largo plazo.")
        elif riesgo == "medio":
            print("IA SIEFORE:")
            print("Se recomienda inversion balanceada.")
        else:
            print("IA SIEFORE:")
            print("Se recomienda inversion conservadora.")

    elif edad >= 30 and edad <= 50:
        if ahorro >= 50000:
            print("IA SIEFORE:")
            print("Buen nivel de ahorro para retiro.")

        else:
            print("IA SIEFORE:")
            print("Necesitas aumentar tus aportaciones.")

        if riesgo == "alto":
            print("Perfil de inversion dinamico.")
        elif riesgo == "medio":
            print("Perfil de inversion equilibrado.")
        else:
            print("Perfil de inversion seguro.")

    else:
        if retiro <= 10:
            print("IA SIEFORE:")
            print("Se recomienda proteger el capital.")
            print("Invertir en opciones seguras.")
        else:
            print("IA SIEFORE:")
            print("Todavia puedes aumentar rendimiento.")

    print("----------------------------------")

    if ahorro < 10000:
        print("Consejo IA:")
        print("Incrementa tu ahorro mensual.")
    elif ahorro >= 10000 and ahorro < 100000:
        print("Consejo IA:")
        print("Tu ahorro es estable.")
    else:
        print("Consejo IA:")
        print("Excelente planeacion financiera.")

    continuar = input("\n¿Deseas realizar otra consulta? (si/no): ").lower()
print("\nGracias por utilizar el sistema SIEFORE")
print("Fin del programa")
print(" Programa Realizado por Zavala Medina NL = 59")
```
